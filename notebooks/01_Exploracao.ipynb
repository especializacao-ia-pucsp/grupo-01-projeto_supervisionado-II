## 9. Análise do cumprimento do SLA

```python
df_chamados['sla_estourado'] = (
    df_chamados['resolution_time_hours'] >
    df_chamados['sla_hours']
)
```

```python
df_chamados['sla_estourado'].value_counts()
```

```python
df_chamados['sla_estourado'].value_counts(normalize=True) * 100
```

A análise identificou que 3.661 chamados (73,22%) ultrapassaram o prazo de SLA, enquanto 1.339 (26,78%) não ultrapassaram esse prazo. O resultado indica uma elevada proporção de chamados com SLA estourado no conjunto de dados, motivando a investigação das taxas por prioridade, departamento e categoria.

### 9.1. Análise do SLA por prioridade

```python
sla_prioridade = df_chamados.groupby('priority')['sla_estourado'].agg(
    quantidade='count',
    sla_estourado='sum',
    percentual_estourado='mean'
)

sla_prioridade['percentual_estourado'] *= 100

sla_prioridade
```

A análise do SLA por prioridade identificou taxas próximas entre os grupos. A prioridade Média apresentou a maior proporção de chamados com SLA estourado (73,80%), enquanto a prioridade Crítica apresentou a menor (72,98%). A diferença de aproximadamente 0,82 ponto percentual indica pouca variação no descumprimento do SLA entre as prioridades.

### 9.2. Análise do SLA por departamento

```python
sla_departamento = df_chamados.groupby('department')['sla_estourado'].agg(
    quantidade='count',
    sla_estourado='sum',
    percentual_estourado='mean'
)

sla_departamento['percentual_estourado'] *= 100

sla_departamento
```

A análise do cumprimento do SLA por departamento apresentou resultados semelhantes entre os setores. Financeiro (73,53%) e TI (73,53%) apresentaram as maiores taxas de chamados com SLA estourado, enquanto Operações apresentou a menor taxa (72,90%), seguida de RH (72,92%). A diferença reduzida entre os departamentos indica pouca variação no descumprimento do SLA entre os setores neste conjunto de dados.

### 9.3. Análise do SLA por categoria

```python
sla_categoria = df_chamados.groupby('target_category')['sla_estourado'].agg(
    quantidade='count',
    sla_estourado='sum',
    percentual_estourado='mean'
)

sla_categoria['percentual_estourado'] *= 100

sla_categoria.sort_values('percentual_estourado', ascending=False)
```

A análise do cumprimento do SLA por categoria identificou diferenças entre os tipos de chamados. Hardware apresentou a maior proporção de chamados com SLA estourado (74,61%), seguido por Software (74,55%). A categoria Acesso apresentou a menor proporção (71,13%). A diferença de aproximadamente 3,48 pontos percentuais entre as maiores e menores taxas indica uma variação relativamente pequena entre as categorias.
